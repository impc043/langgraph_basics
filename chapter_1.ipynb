### Workflow

> let's design basic BMI Calculator </br>
- input(height, weight) --> [bmi-calc] --> output(bmi)

```python
from dotenv import load_dotenv
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
load_dotenv()
```

```python
# step 1 : define the state
class BmiState(TypedDict):
    height: float
    weight: float
    bmi: float
```

```python
# step global: define all your tools
def bmi_calc(state: BmiState) -> BmiState:
    height = state['height']
    weight = state['weight']
    
    state['bmi'] = weight/(height**2)

    return state

```

```python
# step 2: define the graph
graph = StateGraph(BmiState)

# step 3: add nodes to graph
graph.add_node("bmi_calc", bmi_calc)
# graph.add_node("")

# step 4: add edges to graph
graph.add_edge(START, "bmi_calc")
graph.add_edge("bmi_calc", END)

# step 5: compile the graph
workflow = graph.compile()

# step 6: execute the graph
inp_state = {'height': 15.6, 'weight':56}
out_state = workflow.invoke(inp_state)
```
