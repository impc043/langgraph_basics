### Get structure output from model

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from pydantic import BaseModel, Field
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash",
    temperature=0,
    max_tokens=200,
    timeout=None,
    max_retries=1,
    # other params...
)
```

```python
# define your schema that you want from your model instead of default one
class EvaluationSchema(BaseModel):
    feedback: str = Field(description="detail feedback for the essay")
    score: int = Field(description="score the essay on the scale of 0 to 10", ge=0, le=10)
```

```python
# create the instance of structure llm from default llm model
structure_llm = llm.with_structured_output(EvaluationSchema)
```

```python
# your eassy
essay = """  
## Street Dogs
Street dogs are a common sight in many cities and villages around the world. These dogs often live without proper shelter, food, or care. Despite their tough lives, many of them are friendly and intelligent. Street dogs usually depend on leftovers from people or waste bins for food. Unfortunately, they are also exposed to dangers like traffic accidents, diseases, and mistreatment.
Caring for street dogs is important for both animal welfare and public health. Vaccination and sterilization programs help control their population and prevent diseases like rabies. Many animal welfare organizations and volunteers work to rescue, treat, and rehome street dogs.
"""
```

```python
system_prompt = f"Evaluate the following essay for clarity, grammar, coherence, structure, and relevance to the topic in a concise, objective manner. provide the feedback and give score out of 10\n {essay}"
```

```python
# invoke the str-llm
resp = structure_llm.invoke(system_prompt)
```
